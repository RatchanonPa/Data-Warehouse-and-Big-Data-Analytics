```python
from transformers import pipeline

# Load zero-shot classification pipeline
classifier = pipeline("zero-shot-classification", model="MoritzLaurer/multilingual-MiniLMv2-L6-mnli-xnli")

# Example input
text = "สวัสดีครับ คุณพิทธ์ศิริยา ผมชื่อ ภูวนัส เรียกว่า พีท ก็ได้ครับ เป็นที่ปรึกษาจากธนาคารไทยพาณิชย์ ได้รับมอบหมายให้ดูแลด้านการเงินให้กับคุณครับ ก่อนอื่นผมขอแจ้งว่า ผมมีใบอนุญาตแนะนำการลงทุน หมายเลขเจ็ดสามสองสี่สามแปด ซึ่งยังไม่หมดอายุ และมีใบอนุญาตแนะนำประกันวินาศภัยที่ออกโดยหน่วยงานกำกับดูแลนะครับ วันนี้การพูดคุยจะช่วยให้คุณพิศยาได้แนวทางใหม่ในการจัดการการลงทุนรวมถึงวิธีการบริหารความเสี่ยงที่เหมาะสมกับสถานการณ์ปัจจุบันเพื่อช่วยให้คุณบรรลุเป้าหมายทางการเงินได้อย่างมั่นใจครับ ผมขอเวลายี่สิบนาทีในการพูดคุยเรื่องนี้ คุณพิศยาสะดวกไหมครับ"

# Candidate labels (your classes)
candidate_labels = [
    'กล่าวสวัสดี',
    'แนะนำชื่อ และนามสกุล',
    'แจ้งประเภทใบอนุญาต',
    'แจ้งวัตถุประสงค์ของการเข้าพบครั้งนี้',
    'แจ้งประโยชน์ว่าลูกค้าได้ประโยชน์อะไรจากการเข้าพบครั้งนี้',
    'แจ้งระยะเวลาที่ใช้ในการเข้าพบ'
]

# Perform zero-shot classification
result = classifier(text, candidate_labels, multi_label=True)

# Print results
print("Input text:", text)
print("\nClass probabilities:")
for label, score in zip(result['labels'], result['scores']):
    print(f"{label}: {score:.4f}")
```
